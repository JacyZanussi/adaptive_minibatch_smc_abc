Convergence is defined as the last generation such that the acceptance rate < 0.05 for all minibatches considered. For the heterogeneity and physical parameter perturbations, we attempt to generate 20 generations and stop early at the first generation beyond 3600 seconds of wall time. We also stop early when the acceptance rate is less than 0.001(?)

### Hyperparameter sweeps

```python
def get_convergence_index(list_of_series, threshold=0.05):
    # 1. Find the smallest common length (censoring point)
    min_len = min(len(s) for s in list_of_series)
    
    # 2. Search for the first index where ALL meet the condition
    for i in range(min_len):
        if all(series[i] < threshold for series in list_of_series):
            return i
            
    # 3. Fallback: if never met, return the smallest common length
    return min_len-1

def get_last_gen(results, figax = None):
    if type(results) is str:
        data = exps.load(results)
    else:
        data = results
    ar = exps.get_attr(data,'acceptance_rate')
    if figax is None:
        fig,ax = plt.subplots(figsize=(3.5,3))
    else:
        fig,ax = list(figax)
    med_ar = []
    cond = []
    for k in data.keys():
        print(k)
        med = np.median(ar[k],axis=0)
        med_ar.append(med)
        ax.plot(np.log10(med),color = 'cornflowerblue')
    cond = get_convergence_index(med_ar,0.05)
    print(cond)
    choice = np.min(cond)
    print(f'Generation for convergence: {choice + 1}')
    ylim = ax.get_ylim()
    xlim = ax.get_xlim()
    ax.plot([choice,choice],ylim,'green')
    ax.plot(xlim,np.log10([0.05,0.05]),color = 'mediumorchid')
    return fig,ax,choice

get_last_gen('results/lv_constant.pkl')
```

```python
d = exps.load('results/td_constant_scale.pkl')

for k in d.keys():
    print(k)

subset_list = [(128, (2, 2), (10.0, 0.03333333333333333)), (256, (2, 2), (10.0, 0.03333333333333333)), (512, (2, 2), (10.0, 0.03333333333333333))]

dss = exps.subset(d,subset_list)
```

```python
times = exps.get_attr(dss,'total_time')

step_times = {}
for k in times.keys():
    st_list = []
    for rep in range(len(times[k])):
        time_diff = np.diff(times[k][rep])
        st_list.append(time_diff)
    step_times[k] = st_list
```

```python
cmap = plt.get_cmap('viridis')
for k in times.keys():
    fig,ax = plt.subplots()
    for rep in range(len(times[k])):
        steps = step_times[k][rep]
        ax.plot(steps,color = cmap(rep/10))
    
```
