### Estimating thresholds for SMC ABC

The thresholds are obtained as follows:
1. Apply 10 times: full batch SMC ABC (i.e. batch size = sample size) until the signal-to-noise ratio defined as $\frac{\epsilon^2_t}{v_t/n_t} < 2$ or early stopping at an acceptance rate below 0.02, and guaranteeing 5 generations.
2. Calculate the 0.8 quantile of the 10 repeated runs. 0.8 is chosen so that we expect similar convergence for similar runs. 0.5 would place half of these runs a step farther than we want, while 1.0 is subject to some noise. Hence, we take a conservative 0.8.

```python
import experiments 
import numpy as np
```

```python
# Thresholds for the hyperparameter sweep
results_lv = experiments.load('results/lv_lv_threshold.pkl')
results_td = experiments.load('results/td_td_threshold.pkl')

lv_eps = experiments.get_attr(results_lv,'alpha_threshold',trunc=False,slice=-1)
td_eps = experiments.get_attr(results_td,'alpha_threshold',trunc=False,slice=-1)

for k in td_eps.keys():
    print(f'Key: {k}. Epsilon: {np.quantile(td_eps[k],q = 0.8)}')
print(f"Set Epsilon to: {np.quantile(td_eps[k],q = 0.8)} for Transcriptional Dynamics")

for k in lv_eps.keys():
    print(f'Key: {k}. Epsilon: {np.quantile(lv_eps[k],q = 0.8)}')
print(f"Set Epsilon to: {np.quantile(lv_eps[k],q = 0.8)} for Lotka-Volterra")
```

### heterogeneities and physical parameters

```python
# Heterogeneity for transcriptional dynamics.
# heterogeneity are parameters for the gamma and beta distributions centered at 1 0.5.
# We fix their ratio. So, the value corresponds to scale here.
# We pick the largest epsilon, since they are similar. 
results_td_het = experiments.load('results/td_constant_gamma_beta_threshold.pkl')
td_het_eps = experiments.get_attr(results_td_het,'alpha_threshold',trunc=False,slice=-1)
for k in td_het_eps.keys(): 
    print(f'Key: {k}. Epsilon: {np.quantile(td_het_eps[k],q = 0.8)}')
```

```python
# Heterogeneity for Lotka-Volterra. 
# ic_range is the width of the initial condition about 500. IC are uniformly distributed.
# While different, we still pick the largest epsilon, since these runs are MUCH longer.
# Convergence for the pilot threshold estimation is capped at the first generation beyond 900 seconds of wall time (s).
results_lv_het = experiments.load('results/lv_constant_ic_range_threshold.pkl')
lv_het_eps = experiments.get_attr(results_lv_het,'alpha_threshold',trunc=False,slice=-1)
for k in lv_het_eps.keys(): 
    print(f'Key: {k}. Epsilon: {np.quantile(lv_het_eps[k],q = 0.8)}')
```
